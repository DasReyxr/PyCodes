# Sine Wave Visualization
Generate and visualize sine waves: $\sin(t)$ and $\sin(2t)$

## Section 1: Import Required Libraries
Import NumPy for numerical computations and Matplotlib for visualization.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Create time array from 0 to 4π with 1000 points
t = np.linspace(0, 4*np.pi, 1000)
```

```python
# Calculate sine wave values
```

```python
st = 4
sin_t = np.sin(t)
sin_2t = 0.5*np.sin(st*t)

plt.figure(figsize=(12, 6))
plt.plot(t, sin_t, label='signal (f)')
plt.plot(t, sin_2t, label=f'clock ({st}f)')
plt.title(f'k={st}', fontsize=14, fontweight='bold', color='orange')
plt.legend(fontsize=11, loc='upper right')
plt.axis('off')  # hide axis/ticks/frame
plt.tight_layout()
plt.show()
# ...existing code...
```
