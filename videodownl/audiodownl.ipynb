```python
import yt_dlp
from yt_dlp.utils import download_range_func

PATH = r"videos"
DOWNLOAD_AUDIO_ONLY = False  # True para solo audio
TRIMMED = False               # True para recortar el video


def time_to_seconds(t):
    partes = list(map(int, t.split(':')))
    if len(partes) == 2:  # MM:SS
        m, s = partes
        return m * 60 + s
    elif len(partes) == 3:  # HH:MM:SS
        h, m, s = partes
        return h * 3600 + m * 60 + s
    raise ValueError(f"Formato de tiempo inválido: {t}")


def clip_ranges(info_dict, ydl):
    return [{'start_time': time_init, 'end_time': time_end}]


rolas = input("Ingresa el link de la rola: ")

if TRIMMED and not DOWNLOAD_AUDIO_ONLY:
    init = "0:00" #input("Inicio (MM:SS): ")
    end  = "1:58" #input("Fin (MM:SS): ")
    time_init = time_to_seconds(init)
    time_end  = time_to_seconds(end)

COMMON = {
    "cookiesfrombrowser": ("brave",),
    "js_runtimes": {"deno": {}},
    "noplaylist": True,
    "ignoreerrors": False,
    "no_warnings": False,
    "remote_components": ["ejs:github"],
}

if DOWNLOAD_AUDIO_ONLY:
    ydl_opts = {
        **COMMON,
        "format": "bestaudio/best",
        "postprocessors": [{
            "key": "FFmpegExtractAudio",
            "preferredcodec": "mp3",
            "preferredquality": "192",
        }],
        "outtmpl": f"{PATH}/%(title)s.%(ext)s",
    }
elif TRIMMED:
    ydl_opts = {
        **COMMON,
        "format": "bestvideo+bestaudio/best",
        "merge_output_format": "mp4",
        "download_ranges": download_range_func(None, [(time_init, time_end)]),
        "force_keyframes_at_cuts": True,
        "outtmpl": f"{PATH}/%(title)s({init.replace(':', 'm')}-{end.replace(':', 'm')}).%(ext)s",
    }
else:
    ydl_opts = {
        **COMMON,
        "format": "bestvideo+bestaudio/best",
        "merge_output_format": "mp4",
        "outtmpl": f"{PATH}/%(title)s.%(ext)s",
    }


with yt_dlp.YoutubeDL(ydl_opts) as ydl:
    ydl.download([rolas])
```
