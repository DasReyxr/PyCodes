```python
# Compare tan(x) and y=x near zero
x = np.linspace(-0.8, 0.8, 800)

y_tan = np.tan(x)
y_line = x

plt.figure(figsize=(8, 5))
plt.plot(x, y_tan, label='tan(x)', color='tab:blue', linewidth=2)
plt.plot(x, y_line, label='y = x', color='tab:orange', linestyle='--', linewidth=2)
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)
plt.title('tan(x) vs y=x for small x')
plt.xlabel('x (radians)')
plt.ylabel('y')
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

## sinoidal with 4 points

```python
import numpy as np
import matplotlib.pyplot as plt


# 4 muestras
t = np.arange(4)
x = [3, 3, 3, 3]

plt.figure(figsize=(8,4))
plt.plot(t, x, label='Linea continua')

plt.xlabel('Tiempo')
plt.ylabel('Amplitud')
plt.title('Lineas continuas con 4 muestras')
plt.grid(True)
plt.legend()
plt.show()


```
